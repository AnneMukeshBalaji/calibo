````python
import numpy as numpy
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
sns.set_theme(style="whitegrid")
````

````python
df = sns.load_dataset("titanic")
display(df.head())
display(df.info())
display(df.describe())
````

````python
# Step 4
df.isna().sum()

# since only 177 of the 891 rows of age are null we will fill those rows 

# and deck has 688 with null values we will remove it 
````

````python
df.duplicated().sum()
display(df.head())
````

````python
# Step 5: Fill categorical nulls with the most frequent value
display(df['embark_town'].mode()[0])      # check the value first

df['embark_town'] = df['embark_town'].fillna(df['embark_town'].mode()[0])
df['embarked'] = df['embarked'].fillna(df['embarked'].mode()[0])

display(df.head())
display(df.isna().sum())
````

````python
# univariant analysis 
df['survived'].value_counts()
````

````python
sns.countplot(data=df,x="survived")
plt.show()
````

````python
sns.countplot(data=df,x="sex")
plt.show()
````

# step 7 BI-variant Analysis

analysising two variables


````python
fig,ax = plt.subplots(figsize=(7,4))
sns.countplot(data=df,x="sex",hue="survived")
ax.set_title("Gender vs survived")
plt.show()
````

````python
sns.countplot(data=df,x="class",hue="survived",order=['First','Second','Third'])
plt.title("Class Vs Survival")
plt.show()
````
