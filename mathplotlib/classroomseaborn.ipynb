# 1. What is Seaborn?

**Seaborn** is a Python data visualization library built on top of **Matplotlib**.

It is designed especially for:

- Statistical visualization
- Exploratory Data Analysis
- Relationships between variables
- Distribution analysis
- Categorical data analysis
- Correlation analysis

The usual import is:

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

### Simple idea

Matplotlib gives you detailed control over individual plot elements.

Seaborn gives you **high-level statistical plotting functions** that often require less code and produce informative analytical charts quickly.

# 7. Disadvantages

### 1. Less low-level control than Matplotlib

For highly customized figures, Matplotlib may be preferable.

### 2. Can hide complexity

High-level functions are convenient, but beginners may not understand what is happening internally.

### 3. Not a machine learning library

Seaborn only visualizes/analyzes data; it does not train ML models.

### 4. Large datasets can produce crowded plots

Pair plots and categorical plots can become difficult to interpret with too many rows/categories.

### 5. Requires clean/appropriate data

Visualization cannot fix bad data.

### 6. Interactive dashboards are not its main strength

For highly interactive web visualizations, other tools may be more suitable.


# Seaborn – Important Parameters, Concepts, Plots and Their Uses

Seaborn is a Python data visualization library built on top of Matplotlib. It is mainly used for **Exploratory Data Analysis (EDA)** because it provides high-level functions for statistical and categorical visualization with less code.

The following parameters and functions are important for understanding how Seaborn plots work and how to select the correct plot during EDA.

---

## 1. Important Seaborn Parameters and Concepts

| Parameter / Function | Meaning | Commonly Used With | Main Use | Example |
|---|---|---|---|---|
| **`data`** | Specifies the DataFrame containing the data | Almost all Seaborn plots | Tells Seaborn where the data is located | `data=df` |
| **`x`** | Variable displayed on the X-axis | Most plots | Selects the horizontal variable | `x="Study_Hours"` |
| **`y`** | Variable displayed on the Y-axis | Most plots | Selects the vertical variable | `y="Exam_Score"` |
| **`hue`** | Uses different colors for different groups | Scatter, count, bar, box, violin, KDE, line, etc. | Compare categories/groups using color | `hue="Department"` |
| **`style`** | Uses different marker or line styles for groups | Scatter and line plots | Distinguish groups using shape/style | `style="Gender"` |
| **`size`** | Changes marker size according to a variable | Scatter/relational plots | Represent an additional variable | `size="Attendance"` |
| **`palette`** | Controls the color collection used for categories | Categorical/grouped plots | Improve category visualization | `palette="Set2"` |
| **`alpha`** | Controls transparency | Scatter, histogram, KDE, etc. | Handle overlapping observations | `alpha=0.5` |
| **`bins`** | Number of intervals in a histogram | `histplot()` | Control distribution detail | `bins=10` |
| **`fill`** | Fills the area of a distribution | KDE, violin plots | Improve visual readability | `fill=True` |
| **`legend`** | Displays information about colors, markers or sizes | Grouped plots | Explain visual encodings | `legend=True` |
| **`col`** | Creates separate plots horizontally for groups | `relplot()`, `catplot()` and figure-level plots | Compare groups using separate panels | `col="Gender"` |
| **`row`** | Creates separate plots vertically for groups | Figure-level plots | Compare groups using separate panels | `row="Department"` |
| **`kde`** | Adds a Kernel Density Estimate curve | `histplot()` | Show smooth distribution shape | `kde=True` |

# 15. Univariate, Bivariate and Multivariate Analysis — Mandatory Reference

This classification should be understood **before learning individual plots**. The number of variables determines the analytical question, while the data type and purpose determine the appropriate plot.

## Univariate Analysis

**One variable at a time.**

| Plot | Typical Variable | Main Use | Student Dataset Example |
|---|---|---|---|
| Histogram | One numerical | Frequency, distribution shape, skewness | `Exam_Score` distribution |
| KDE | One numerical | Smooth density estimate | `Study_Hours` distribution |
| ECDF | One numerical | Cumulative distribution | Cumulative `Exam_Score` |
| Rug | One numerical | Individual observations | `Study_Hours` values |
| Count Plot | One categorical | Category frequency | Students by `Department` |
| Bar Plot | Category + aggregation | Compare aggregated values | Mean `Exam_Score` by `Department` |
| Box Plot | One numerical | Spread, median, potential outliers | `Exam_Score` |
| Violin Plot | One numerical | Distribution shape and spread | `Attendance` |

### Main purposes

- Understand distribution
- Understand frequency
- Identify spread
- Identify central tendency
- Investigate potential outliers
- Understand category balance

## Bivariate Analysis

**Two variables together.**

| Plot | Variable Combination | Main Use | Student Dataset Example |
|---|---|---|---|
| Scatter Plot | Numerical + Numerical | Relationship, trend, clusters | `Study_Hours` vs `Exam_Score` |
| Line Plot | Ordered/time + Numerical | Trend over an ordered axis | Average `Exam_Score` by `Enrollment_Date` |
| Bar Plot | Categorical + Numerical | Group comparison | Mean `Exam_Score` by `Department` |
| Box Plot | Categorical + Numerical | Compare distributions and outliers | `Exam_Score` by `Department` |
| Violin Plot | Categorical + Numerical | Compare distribution shapes | `Attendance` by `Department` |
| Strip Plot | Categorical + Numerical | Display individual observations | `Exam_Score` by `Placement_Status` |
| Swarm Plot | Categorical + Numerical | Individual observations with reduced overlap | `Exam_Score` by `Department` |
| Regression Plot | Numerical + Numerical | Relationship plus fitted trend | `Study_Hours` vs `Exam_Score` |

### Main purposes

- Compare two variables
- Identify relationships
- Detect trends
- Compare groups
- Investigate possible correlation

## Multivariate Analysis

**Three or more variables or multiple dimensions of information.**

| Plot / Technique | Additional Dimension | Main Use | Student Dataset Example |
|---|---|---|---|
| Scatter + `hue` | Category | Compare numerical relationship across groups | Study Hours + Exam Score + Department |
| Scatter + `style` | Category | Use marker style for another grouping | Study Hours + Exam Score + Gender |
| Scatter + `size` | Numerical | Encode magnitude with point size | Study Hours + Exam Score + Attendance |
| Pair Plot | Many numerical variables | Pairwise relationship exploration | Attendance, Study Hours, Assignment Score, Exam Score |
| Heatmap | Many numerical variables | Correlation/matrix overview | Numerical feature correlations |
| Facet Plot | Category | Repeat analysis across groups | Study Hours vs Exam Score by Department |
| Grouped Bar Plot | Category + category + numerical | Compare aggregated values across two categories | Exam Score by Department and Placement Status |

### Main purposes

- Understand interactions
- Compare subgroups
- Detect patterns hidden in simpler charts
- Examine multiple features together
- Support feature understanding before ML

# 16. Plot Selection Guide — When, Why and Purpose

| Plot | When to Use | Main Purpose | Key Interpretation |
|---|---|---|---|
| **Histogram** | One numerical variable | Distribution and frequency | Which ranges contain most observations? |
| **KDE** | Smooth numerical distribution needed | Density estimation | Where are observations concentrated? |
| **ECDF** | Cumulative behavior is important | Cumulative distribution | What proportion is at or below a value? |
| **Count Plot** | One categorical variable | Frequency comparison | Which categories are most/least frequent? |
| **Bar Plot** | Category with an aggregated numerical measure | Category comparison | Which group has the higher aggregate value? |
| **Scatter Plot** | Two numerical variables | Relationship analysis | Is there a visible trend, cluster or outlier? |
| **Line Plot** | X-axis has meaningful order/time | Trend analysis | How does the measure change across the ordered axis? |
| **Box Plot** | Grouped numerical distributions | Spread and outlier analysis | How do median, spread and potential outliers differ? |
| **Violin Plot** | Grouped distribution shape matters | Distribution comparison | Where is each group's density concentrated? |
| **Strip Plot** | Individual observations matter | Sample-level comparison | Where do individual observations lie? |
| **Swarm Plot** | Moderate sample size | Individual observations with reduced overlap | How are observations distributed within categories? |
| **Regression Plot** | Two numerical variables with a trend question | Fitted relationship | Is a simple trend line a useful summary? |
| **Joint Plot** | One numerical pair needs joint + marginal views | Combined relationship/distribution | What do the relationship and marginal distributions show together? |
| **Pair Plot** | Several numerical features | Rapid pairwise EDA | Which feature pairs show interesting relationships? |
| **Heatmap** | Matrix/correlation values | Compact comparison | Which variable pairs show stronger positive/negative association? |
| **Facet Plot** | Same relationship across meaningful groups | Group-specific comparison | Does the relationship change between groups? |

# 17. Seaborn Plot Families and Major Functions

Seaborn does **not have one fixed official number of plots**. It provides a collection of plotting functions. The following families cover the major functions students should know for EDA.

| Plot Family | Major Functions | Typical Analysis |
|---|---|---|
| Relational | `scatterplot()`, `lineplot()` | Bivariate and multivariate relationships |
| Distribution | `histplot()`, `kdeplot()`, `ecdfplot()`, `rugplot()` | Univariate distributions |
| Categorical | `barplot()`, `countplot()`, `boxplot()`, `violinplot()`, `stripplot()`, `swarmplot()`, `pointplot()` | Category comparison and distributions |
| Regression | `regplot()`, `lmplot()` | Relationship and regression trend |
| Matrix | `heatmap()`, `clustermap()` | Correlation and matrix analysis |
| Pairwise | `pairplot()`, `PairGrid()` | Multiple numerical relationships |
| Joint | `jointplot()` | Bivariate relationship + marginal distributions |
| Faceting | `FacetGrid()`, `relplot()`, `catplot()` | Repeated analysis across groups |
| Objects interface | `seaborn.objects` | Composable modern visualization interface |

The exact API can vary by installed Seaborn version. Students should check the version when reproducing notebooks in another environment.

# 16. Univariate — Count Plot

Use `countplot()` for a categorical variable.

A count plot in Seaborn is used to count how many times each category appears in a column and display those counts as bars.

Question:

> How many students belong to each department?


````python
plt.figure(figsize=(7, 4))

sns.countplot(
    data=df,
    x="Department"
)

plt.title("Student Count by Department")
plt.xlabel("Department")
plt.ylabel("Number of Students")

plt.show()
````

# 17. Univariate — Histogram

Use `histplot()` for a numerical variable.

Question:

> How are attendance values distributed?


````python
# Kernel Density Estimate-A smooth version of a histogram that shows where the data is concentrated.
plt.figure(figsize=(8, 4))

sns.histplot(
    data=df,
    x="Attendance",
    bins=8,
    kde=True
)

plt.title("Attendance Distribution")
plt.xlabel("Attendance (%)")
plt.ylabel("Number of Students")

plt.show()
````

# 27. Bivariate — Regression Plot

`regplot()` can show a scatter relationship with a fitted regression line.

A regression plot in Seaborn is used to show the relationship between two numerical variables and draw a regression line that represents the overall trend.

Question:

> Is there a visible linear relationship between study hours and assignment score?

Important:

**A regression line in a visualization does not automatically mean that the relationship is causal.**

It creates:

Scatter points → actual data

Regression line → best-fit linear relationship

Confidence interval → shaded region around the line


````python
plt.figure(figsize=(8, 5))

sns.regplot(
    data=df,
    x="Study_Hours",
    y="Assignment_Score"
)

plt.title("Study Hours vs Assignment Score with Trend")
plt.xlabel("Study Hours")
plt.ylabel("Assignment Score")
plt.tight_layout()
plt.show()
````

# 33. Multivariate — Pair Plot

`pairplot()` is one of the most useful EDA functions.

It creates pairwise relationships among numerical variables.

For our dataset:

- Age
- Attendance
- Assignment Score
- Study Hours

It can also use `hue` for category grouping.

### Important

Pair plots can become crowded when there are many features. Use a carefully selected subset.

Pairplot = distribution + relationship


````python
numeric_cols = [
    "Age",
    "Attendance",
    "Assignment_Score",
    "Study_Hours"
]

pair_df = df[numeric_cols + ["Department"]].dropna()

sns.pairplot(
    pair_df,
    hue="Department"
)

plt.show()
````

# 34. Multivariate — Correlation Heatmap

A heatmap represents a matrix using color intensity.

A correlation heatmap helps answer:

> Which numerical variables move together?

Correlation values generally range from:

```text
-1 → strong negative relationship
 0 → little/no linear relationship
+1 → strong positive relationship
```

Important:

**Correlation does not prove causation.**


````python
numeric_cols = [
    "Age",
    "Attendance",
    "Assignment_Score",
    "Study_Hours"
]

corr = df[numeric_cols].corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    vmin=-1,
    vmax=1
)

plt.title("Correlation Heatmap")
plt.show()
````

# 38. Multivariate — Faceting

Faceting means creating separate plots for different subsets of data.

Seaborn provides functions such as `relplot()` and `catplot()`.

Example:

- Separate panel for each gender
- Same relationship in each panel



````python
#relplot() is mainly used when you want to understand the relationship between numerical variables.
"""It works with:

scatter → relationship between two numerical variables
line → trend over an ordered variable"""


sns.relplot(
    data=df,
    x="Study_Hours",
    y="Assignment_Score",
    col="Gender",
    hue="Department",
    kind="scatter",
    height=4,
    aspect=1
)

plt.show()
````

````python
#catplot() is mainly used when one or both variables are categorical.
"""it works with strip, swarm, box, violin, boxen, point, bar, count """
sns.catplot(
   data=df,
   x="Department",
   y="Assignment_Score",
   col="Gender",
   hue="Department",
   kind="violin",
   height=4,
   aspect=1
)

plt.show()
````

# 51. Common Seaborn Mistakes

## Mistake 1 — Using the wrong analysis type

Do not call everything multivariate.

- One variable → univariate
- Two variables → bivariate
- Three or more → multivariate

## Mistake 2 — Confusing countplot and barplot

`countplot()` counts observations.

`barplot()` normally displays an aggregated numerical statistic.

## Mistake 3 — Using too many variables

Adding `hue`, `style`, `size`, `row`, and `col` all at once can make a plot impossible to understand.

## Mistake 4 — Treating correlation as causation

A strong correlation does not prove that one variable causes another.

## Mistake 5 — Plotting dirty data

Clean and validate numerical/categorical values first.

## Mistake 6 — Ignoring missing values

Missing values can change distributions.

## Mistake 7 — Using pairplot on every column

Pair plots should usually use a carefully selected set of numerical features.

## Mistake 8 — Overusing KDE

KDE is an estimate, not the original data itself.

## Mistake 9 — Ignoring sample size

A complex plot can be misleading when there are very few observations.

## Mistake 10 — Beautiful chart, no question

Every EDA plot should answer a question.


# 52. Common AI/ML Mistakes

### 1. Looking at the target after preprocessing incorrectly

Be careful not to introduce leakage when preparing features.

### 2. Selecting features because a chart looks interesting

Feature selection should use domain knowledge, validation and appropriate statistical/ML methods.

### 3. Assuming correlation means usefulness

A feature may have nonlinear relationships that correlation does not capture.

### 4. Removing outliers automatically

Investigate them first.

### 5. Using all features in pairplot

This can produce an unreadable visualization.

### 6. Interpreting visual patterns as proof

EDA helps generate hypotheses. It does not by itself establish causation.

The same analytical task is implemented using several open-source Python visualization libraries. This demonstrates that the **analytical question stays the same while the programming interface changes**.

**Compare the average `Exam_Score` across `Department`.**

Libraries:

2. Seaborn
3. Pandas Plotting
4. Plotly
5. Bokeh
6. Altair
7. Plotnine

The interactive appearance depends on the notebook environment.

````python
department_mean = (
    df.groupby("Department", as_index=False)["Exam_Score"]
      .mean()
)

department_mean
````

## Matplotlib

**Style:** lower-level, highly customizable static visualization.

**Strength:** precise control over chart components.

````python
plt.figure(figsize=(9, 5))
plt.bar(department_mean["Department"], department_mean["Exam_Score"])
plt.title("Average Exam Score by Department — Matplotlib")
plt.xlabel("Department")
plt.ylabel("Average Exam Score")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
````

## Seaborn

**Style:** high-level statistical visualization.

**Strength:** concise DataFrame-oriented EDA syntax.

````python
plt.figure(figsize=(9, 5))
sns.barplot(data=df, x="Department", y="Exam_Score", estimator="mean")
plt.title("Average Exam Score by Department — Seaborn")
plt.xlabel("Department")
plt.ylabel("Average Exam Score")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
````

## Pandas Plotting

**Style:** quick DataFrame plotting.

**Strength:** convenient for rapid exploratory charts.

````python
ax = (df.groupby("Department")["Exam_Score"].mean()).plot(
    kind="bar", figsize=(9, 5), title="Average Exam Score by Department — Pandas"
)
ax.set_xlabel("Department")
ax.set_ylabel("Average Exam Score")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
````
