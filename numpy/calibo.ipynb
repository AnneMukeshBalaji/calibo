```python
print("Hello World")

s = "Hello World"
# reverse of a string witout any predefined functions
def reverse(s:str) -> str :
  reverse_str = ""
  for i in range(len(s)-1,-1,-1):
    reverse_str += s[i]
  return reverse_str
print(reverse(s))
```
