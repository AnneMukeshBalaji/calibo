```python
print("Hello World")

s = "Hello World"
# reverse of a string witout any predefined functions
def reverse(s:str) -> str :
    reverse_str = ""
    for i in range(len(s)-1,-1,-1):
        reverse_str += s[i]
    return reverse_str
print(reverse(s))
```

```python

# NOTE: BroadCasting In Numpy
# DEF : BroadCasting allows numpy to perform opertaions between arrays with compatible shapes 
# Adding matrix with vector 
matrix = np.array([[1,2,3],[4,5,6]])
vector = np.array([10,20,30])
print("Matrix + Vector : ",matrix+vector)
```

```python
# NOTE: Axis tells numpy which dimesions to operate across 
# axis = 1 -> operates down rows,
# axis = 0 -> operates down cols 
example_matrix = np.array([[10,20,30],[40,50,60]])
print("Row Sum : ",np.sum(example_matrix,axis=1))
print("Columns sum : ",np.sum(example_matrix,axis=0))
```

```python
# NOTE: Filtering using where() method 
example_vector = np.array([1,2,3,4,5,20,21,22,23,24,25])
print("Elements Greather that 20 : ",example_vector > 20) # It Returns boolen result array 
print("Values : ",example_vector[example_vector > 20])
# Using Where Without user defined message -> it returns only the  index of the elements that satsified the condition 
print("Using Where Method Without Any User Definied Message : ",np.where(example_vector > 20))
print("Using Where Method With User Definied Message : ",np.where(example_vector > 20,"Yes","No"))
```

```python
# NOTE: Combining Arrays 
# Some common methods include as : 
# 1] np.concatenate()
# 2] np.vstack()
# 3] np.hstack()
# 4] np.stack() -> creates a new axis 

matrix_1 = np.array([[1,2,3],[4,5,6]])
matrix_2 = np.array([[6,7,8],[9,10,11]])
print("Concatenation : ",np.concatenate([matrix_1,matrix_2]))
print("VStack : ",np.vstack([matrix_1,matrix_2]))
print("HStack : ",np.hstack([matrix_1,matrix_2]))
print("Stack : ",np.stack([matrix_1,matrix_2],axis=1))
print("Stack : ",np.stack([matrix_1,matrix_2],axis=0))
```

```python
# NOTE: Random number generation
rng = np.random.default_rng(42) 
print("Random Integers : ",rng.integers(1,10,size=5))
print("Random Floats : ",rng.random(5))
print("Normal Distribution : ",rng.normal(size=5))
```

```python
# NOTE: Different types of multiplication 
# * -> it does element multiplication 
# @ -> it does rows multiplication 
# .dot() method does dot product between vectors only 
matrix_1 = np.array([[1,2],[3,4],[5,6]]) # (3,2)
matrix_2 = np.array([[6,7,8],[9,10,11]]) # (2,3)
print("Matrix Multiplication : ")
print(matrix_1 @ matrix_2) # (2 == 2) so matrix multiplication is possible 
matrix_1 = np.reshape(matrix_1,(2,3))
print("After Reshaping Matrix 1 : ")
print(matrix_1)
print("Element Multiplication : ")
print(matrix_1 * matrix_2)
vector_1 = np.array([1,2,3,4,5])
vector_2 = np.array([6,7,8,9,10])
print("Dot Product : ",vector_1 * vector_2)
```

```python
# Nan Safe mean and sum 
nan_arr = np.array([10,20,30,np.nan,40,50])
print("Is Nan : ",np.isnan(nan_arr))
print("Mean Value : ",np.mean(nan_arr))
print("Nan Safe Mean : ",np.nanmean(nan_arr))
print("Sum : ",np.sum(nan_arr))
print("Nan Safe Sum : ",np.nansum(nan_arr))
```
