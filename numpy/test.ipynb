```python
#11. Filter all rows where the first feature is greater than 50.
X = np.array([[10, 20, 30],[60, 70, 80],[40, 50, 60],[90, 10, 20],[15, 25, 35]])
print("Filtered rows where first feature > 50: ",X[X[:,0] > 50])
```

```python
#12. Standardize every feature column.
mean = np.mean(X, axis=0)
std = np.std(X, axis=0)
X_standardized = (X - mean) / std
print("Mean of each feature column :", mean)
print("Std  of each feature column :", std)
print("Standardized Feature Matrix :\n", X_standardized)
```

```python
#13. Create a random matrix of shape (10, 4) using a fixed seed.
np.random.seed(42)
random_matrix = np.random.randn(10, 4)
print("Random Matrix of shape (10, 4):\n", random_matrix)
print("Shape :", random_matrix.shape)
```

```python
#14. Calculate X @ weights for a feature matrix and weight vector.
np.random.seed(42)
X_feat = np.random.randn(10, 4)
weights = np.array([0.5, -0.3, 0.8, 1.2])
result = X_feat @ weights
print("Feature Matrix X_feat shape :", X_feat.shape)
print("Weights                     :", weights)
print("X @ weights                 :", result)
print("Result Shape                :", result.shape)
```

```python
#15. Add a bias vector using broadcasting.
bias = np.array([1.0, 2.0, 3.0, 4.0])
X_biased = X_feat + bias
print("Bias vector             :", bias)
print("X_feat + bias (first 3 rows):\n", X_biased[:3])
print("Shape after adding bias :", X_biased.shape)
```

```python
#16. Calculate MSE between y_true and y_pred.
y_true = np.array([3.0, -0.5, 2.0, 7.0])
y_pred = np.array([2.5,  0.0, 2.0, 8.0])
mse = np.mean((y_true - y_pred) ** 2)
print("y_true                   :", y_true)
print("y_pred                   :", y_pred)
print("Mean Squared Error (MSE) :", mse)
```

```python
#19. Create an array containing NaN and calculate its NaN-safe mean.
arr_with_nan = np.array([1.0, 2.0, np.nan, 4.0, 5.0, np.nan])
print("Array with NaN   :", arr_with_nan)
print("Normal mean      :", np.mean(arr_with_nan))     # returns nan
print("NaN-safe mean    :", np.nanmean(arr_with_nan))  # ignores NaN
```

```python
#20. Explain the difference between A * B and A @ B in your own words.
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
print("A =\n", A)
print("B =\n", B)
print()
print("A * B  (Element-wise multiplication):\n", A * B)
print()
print("A @ B  (Matrix multiplication / dot product):\n", A @ B)
print()
print("""
Difference:
  A * B  → Element-wise multiplication.
            Each element of A is multiplied by the SAME-POSITION element of B.
            Example: A[0][0] * B[0][0] = 1 * 5 = 5
            Both arrays must have the same shape.

  A @ B  → Matrix multiplication (dot product).
            Row of A is multiplied by column of B and the results are summed.
            Example: A[0] · B[:,0] = (1*5) + (2*7) = 19
            Shape rule: (m x n) @ (n x p) = (m x p)
""")
```
